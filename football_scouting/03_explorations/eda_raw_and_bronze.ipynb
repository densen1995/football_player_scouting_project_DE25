# EDA RAW

## How many files did the ingestion notebook create per type?

```python
import json

VOLUME_PATH = "/Volumes/football_scouting/default/raw"

spark.sql(f"LIST '{VOLUME_PATH}'").display()

for entity in ["leagues", "teams", "players", "fixtures"]:
    files = spark.sql(f"LIST '{VOLUME_PATH}/{entity}'").count()
    print(f"{entity}: {files} files")
```

# EDA BRONZE

## Look at one raw API record(A random player search)

```python
record = spark.sql("SELECT raw_json FROM football_scouting.bronze.raw_players LIMIT 1").first()
print(json.dumps(json.loads(record["raw_json"]), indent=2)[:3000])
```

## Check which seasons i have 

```python
%sql


SELECT season, league_id, COUNT(*) AS players
FROM football_scouting.bronze.raw_players
GROUP BY season, league_id
ORDER BY season, league_id;
```

## A player from a specific season (and league)

```python

record = spark.sql("""
    SELECT raw_json FROM football_scouting.bronze.raw_players
    WHERE season = 2026 AND league_id = 39
    LIMIT 1
""").first()

print(json.dumps(json.loads(record["raw_json"]), indent=2)[:3000])
```
