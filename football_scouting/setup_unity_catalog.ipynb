# Setup Unity Catalog 
Creates the catalog, the schemas for the medallion layers and the Volume where the API files land.

```python
%sql
CREATE CATALOG IF NOT EXISTS football_scouting;

USE CATALOG football_scouting;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

CREATE VOLUME IF NOT EXISTS football_scouting.default.raw;
```

```python
# One folder per type of data. The bronze streaming tables read from these folders.
dbutils.fs.mkdirs("/Volumes/football_scouting/default/raw/leagues")
dbutils.fs.mkdirs("/Volumes/football_scouting/default/raw/teams")
dbutils.fs.mkdirs("/Volumes/football_scouting/default/raw/players")
dbutils.fs.mkdirs("/Volumes/football_scouting/default/raw/fixtures")

# The ingestion notebook remembers here what it already downloaded
dbutils.fs.mkdirs("/Volumes/football_scouting/default/raw/_progress")
```
